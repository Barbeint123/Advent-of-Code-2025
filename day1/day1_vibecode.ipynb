```python
# T1

pointer = 50
password = 0

with open("day1.txt", encoding="utf-8") as f:
    for line in f:
        sign = 1 if line[0] == "R" else -1
        steps = int(line[1:])

        pointer = (pointer + sign * steps) % 100
        password += pointer == 0

print(password)
```

```python
# T2

pointer = 50
password = 0

with open("day1.txt") as f:
    for line in f:
        sign = 1 if line[0] == "R" else -1
        steps = int(line[1:])

        old = pointer
        pointer += sign * steps
        password += abs(pointer // 100) - abs(old // 100)
        pointer %= 100

print(password)
```
